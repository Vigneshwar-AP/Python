## User define function with for loop

#### Example1:Get a multiple input for the list

```python
def list_input(length):
    list_1 = []
    for i in range(0,length):
        Input_value=input("Enter the value you need to enter in the list:")
        list_1.append(Input_value)
    return list_1
list_input(int(input("enter the length of the list you need:")))
```

#### Example2:Factorial of number

```python
def factorial(fact_num):
    for i in range(1,fact_num):
        fact_num=fact_num*i
    return fact_num
factorial(3)
        
```

## User define function with for loop, if condition

#### Example 1:Find the greatest number in the list

```python
def list_input(length):
    list_1 = []
    for i in range(0,length):
        Input_value=int(input("Enter the value you need to enter in the list to find greatest number:"))
        list_1.append(Input_value)
    return  list_1
my_list=list_input(int(input("enter the length of the list you need:")))
def greatest_number(my_list):
    greatest=my_list[0]
    for i in range(0,len(my_list)):
        if greatest<my_list[i]:
            greatest=my_list[i]
    return "The greatest number in your list is:",greatest
greatest_number(my_list)
```

#### Example 2: Print sum of odd numbers in list

```python
def sum_of_odd_number(number_list):
    average=0
    for i in range(0,len(number_list)):
        if number_list[i]%2 != 0:
            average=average+number_list[i]
    return "Average of odd numbers in list is:",average
number_list=[1,2,3,4,5,6,7,8,9]
sum_of_odd_number(number_list)
```

## Break

#### Example 1

```python
sample_list=[]
while True:
    name=input("Enter the name you want to add in list:")
    print("If you want to break the list give ',' to break")
    if name==',':
        break
    sample_list.append(name)
print(sample_list)
```

## Continue

#### Example 1: Skip the threes

```python
number_list=[99,18,3,2,5,6,121,8,9,120]
average=0
for i in range(0,len(number_list)):
    if number_list[i]%3==0:
        print("{} is divisible by 3 so we are skipping this".format(number_list[i]))
        continue
    else:
        average=average+number_list[i]
print(average)
```

#### Example 2: Removing vowels in Sentence

```python
string='Hi I am vigneshwar'
converting_to_lower=string.lower()
for letter in converting_to_lower:
    if letter=='a' or letter=='e' or letter=='i' or letter=='o' or letter=='u':
        continue
    else:
        print(letter)
```

## For loop with Not In operator

#### Example 1

```python
all_employees=['ramesh','suresh','kamesh','vimal','somu']
employee_termination_list=['ramesh','somu']
employee_going_to_retained=[]
for user in all_employees:
    if user not in employee_termination_list:
        employee_going_to_retained.append(user)
print(employee_going_to_retained)
```

#### Example 2

```python
all_values=[99,99,27,36,87,77,77,25,27]
original_list=[]
for value in all_values:
    if value not in original_list:
        original_list.append(value)
print(original_list)
```

## For loop with In operator

#### Example 1

```python
all_values=[99,27,36,87,77,25,27]
original_list=[27,99,87]
final_list=[]
for value in all_values:
    if value in original_list:
        final_list.append(value)
print(final_list)
```

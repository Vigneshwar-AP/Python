## First Example : Creating Nick Name

### Use case:
#### Get two user input from the user (First name, Last name)
#### Capitalizing the both first and last name
#### Concatinating the first character of first and last name alias nick name
#### Verifying whether nick_name is valid by returning boolean value as True

```python
print("Give your First and Last name as Input")
First_name=input("Enter your First Name:")
Last_name=input("Enter your Last Name:")

print(".............Capitalizing the both first and last name.................")
Capital_First_name=First_name.upper()
Capital_Last_name=Last_name.upper()
print(Capital_First_name,Capital_Last_name)
print(".............Concatinating the first character of first and last name alias nick name.................")
nick_name=First_name[0]+Last_name[0]
print(nick_name)
print("...........convert the nick name to lower case...................")
Lower_nick_name=nick_name.lower()
print(Lower_nick_name)
First_name_lower=First_name.lower()
Last_name_lower=Last_name.lower()
print(".........Verifying whether nick_name is valid by returning boolean value as True....................")
print(First_name_lower.startswith(Lower_nick_name[0]))
print(Lower_nick_name.endswith(Last_name_lower[0]))
```

## Second Example: Feedback form

### Use Case: Enter the details asking in below form

```python
print("Enter the details to fill the feedback form for your course you got tutored")
Title=input("Enter the course name:")
print(Title.title())
Name=input("Enter the name:")
print(Name.capitalize())
Ph_no=input("Enter the phone no:")
print("Phone number format which was entered is",Ph_no.isdigit())
Address=input("Enter your State First letter along with pincode:")
print("Address format which you entered is ",Address.isalnum())
feedback=input("Enter the feedback for the course:")
print(feedback.strip())
print("Try to replace any word in feedback, enter the word to be replaced first and then new word to be entered:")
print(feedback.replace(input(),input()))
print("Thankyou {} for your valuble feedback to your {} course".format(Name,Title))
```

## Mini Hangman

```python
import random
Game_list=["python","machine","learning","bridge"]
word=random.choice(Game_list)
length=len(word)
List_word=list(word)
List1=["_"]*length
s="__________"
i=1
while True:
     found=False
     if "".join(List1)==word:
            print("Congrats you found the word.")
            break

     print("_______ guess the letter")
     guessed_letter=input("Enter the letter you are gussing")
    
     if guessed_letter.lower() in word.lower():
        print("Hurray, your guess is right")
        for index, char in enumerate(word):
            if char==guessed_letter:
                List1[index]=guessed_letter
        found=True
     elif i==6:
            print("You exceeded the gussing limit")
            print("The word you missed to guess is {}".format(word))
            break  
     elif found==False:
         print("your guess is wrong. Attempt {}".format(i))
         i=i+1

print("".join(List1))
```

### 2.	Create a list of numbers and print only even numbers using a for loop.

```python
numbers_list=[3,22,321,334,556,75,34,68,222,1234,55,57,89,97]
for i in range(0,len(numbers_list)):
    if numbers_list[i]%2==0:
        print("{} is a even number".format(numbers_list[i]))
```

### 3.	Write a program that handles division by zero using try-except.

```python
divident=int(input("Enter the divident"))
divisor=int(input("Enter the divisor"))

try:
    print(divident/divisor)
except ZeroDivisionError:
    print("Invalid number!, Enter a valid number")
```

### 4.	Given a string, count how many vowels it has.

```python
string= "Hi I am Vigneshwar"
count=0
for vowels in string:
    if vowels.lower() in 'aeiou':
        print(vowels)
        count+=1
print("Total vowels count is {}".format(count))
```

### 5.	Reverse a list without using built-in functions.

```python
list1=[1,2,3,4,5]
list2=[]
for value in reversed(list1):
    list2.append(value)
print(list2)
```

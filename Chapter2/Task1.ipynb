```python
company={'Microsoft','apple','google','banana','guva'}
company2={'cherry','google','apple','Microsoft'}
output3=company2.intersection(company)
print(output3)
company.intersection_update(company2)
print(company)
output=company.symmetric_difference(company2)
print(output)
company2.symmetric_difference_update(company)
print(company2)
output2=company2.difference(company)
print(output2)
company.difference_update(company2)
print(company)
```

```python
tf=Sample_list4.isdisjoint(company)
print(tf)
compare1={'Microsoft','apple','google','banana','guva'}
compare2={'google','apple','Microsoft'}
TF=compare2.issubset(compare1)
print(TF)
FT=compare1.issuperset(compare2)
print(FT)
```

## Rock,Paper,Scissor game

```python
import random

print("""Enter any value from these three
          Rock
          Paper
          Scissor""")
game_list=['rock','paper','scissor']
while True:
    value=random.choice(game_list)
    user_value=input("Enter the value you are going to guess:")
    if value.lower()==user_value.lower():
       print("Computer choose {}".format(value))
       print("Your enterd same as computer. Try again!")
    elif (user_value.lower()=="rock" and value.lower()=="paper") or (user_value.lower()=="paper" and value.lower()=="scissor") or (user_value.lower()=="scissor" and value.lower()=="rock"):
         print("Computer choose {}".format(value))
         print("Computer wins. Try again!")
    elif (user_value.lower()=="paper" and value.lower()=="rock") or (user_value.lower()=="scissor" and value.lower()=="paper") or (user_value.lower()=="rock" and value.lower()=="scissor"):
        print("Computer choose {}".format(value))
        print("Hurray you win.")
    redo=input("If you want to guess again type yes or else type no")
    if redo.lower()=='yes':
        continue
    else:
        break
```
